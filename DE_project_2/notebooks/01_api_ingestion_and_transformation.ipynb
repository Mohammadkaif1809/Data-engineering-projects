```python
carts_url = 'https://dummyjson.com/carts?limit=0'
```

```python

carts_response = requests.get(carts_url)
carts_response.raise_for_status()

carts_data = carts_response.json()
```

```python

carts_data
```

```python
carts_df = pd.DataFrame(carts_data['carts'])
```

```python
carts_df
```

```python
carts_df['products'][0]
```

```python
carts_products_data = []

for _,row in carts_df.iterrows():
  cart_id = row['id']
  for product in row['products']:
    carts_products_data.append({
      "cart_id" : cart_id,
      'product_id': product['id'],
      'title': product['title'],
      'price': product['price'],
      'quantity': product['quantity'],
      'total': product['total'],
      'discountPercentage': product['discountPercentage'],
      'discountedTotal': product['discountedTotal'],
      'thumbnail': product['thumbnail']
    })
    
dim_carts_products = pd.DataFrame(carts_products_data)
dim_carts_products
```

```python
carts_df = carts_df.drop(columns=['products'])
```
