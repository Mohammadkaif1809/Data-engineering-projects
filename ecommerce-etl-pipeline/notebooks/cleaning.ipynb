```python
def correct(x):
  word = ""
  for ch in x:
    if ch not in " .,!@#$%^&*()_-+=":
      word += ch.lower()
      
      if word == "cod":
            return "cash on delivery"
          
      if word in ("net","credit","debit","cash") or word.endswith("on"):
        word += " "
  return word
```

```python
df['Payment_Method'].apply(correct).unique()
```

```python
df['Payment_Method'] = df['Payment_Method'].apply(correct)
```
